### Database connected

```python
import sqlite3

conn = sqlite3.connect("cricket.db")
cursor = conn.cursor()

print("Database connected successfully!")
```

### Team table

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS teams (
    team_id INTEGER PRIMARY KEY AUTOINCREMENT,
    team_name TEXT NOT NULL,
    captain TEXT
)
""")

conn.commit()

print("Teams table created!")
```

### Player Table

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS players (
    player_id INTEGER PRIMARY KEY AUTOINCREMENT,
    player_name TEXT NOT NULL,
    team_id INTEGER,
    role TEXT,
    jersey_no INTEGER,
    FOREIGN KEY(team_id) REFERENCES teams(team_id)
)
""")

conn.commit()

print("Players table created!")
```

### Match Table

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS matches (
    match_id INTEGER PRIMARY KEY AUTOINCREMENT,
    team1 TEXT,
    team2 TEXT,
    date TEXT,
    venue TEXT,
    toss_winner TEXT,
    batting_team TEXT,
    score1 INTEGER,
    wickets1 INTEGER,
    score2 INTEGER,
    wickets2 INTEGER,
    winner TEXT
)
""")

conn.commit()

print("Matches table created!")
```

### Point Table 

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS points (
    team_name TEXT PRIMARY KEY,
    played INTEGER DEFAULT 0,
    won INTEGER DEFAULT 0,
    lost INTEGER DEFAULT 0,
    points INTEGER DEFAULT 0
)
""")

conn.commit()

print("Points table created!")
```

### Statistics Table

```python
import sqlite3

# ==============================
# CRICKET TOURNAMENT SYSTEM
# ==============================

conn = sqlite3.connect("cricket_tournament.db")
cursor = conn.cursor()

# ---------- DATABASE TABLES ----------

cursor.execute("""
CREATE TABLE IF NOT EXISTS tournament (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT,
    overs INTEGER
)
""")

cursor.execute("""
CREATE TABLE IF NOT EXISTS teams (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT UNIQUE,
    captain TEXT
)
""")

cursor.execute("""
CREATE TABLE IF NOT EXISTS players (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT,
    team_id INTEGER,
    role TEXT,
    runs INTEGER DEFAULT 0,
    wickets INTEGER DEFAULT 0
)
""")

cursor.execute("""
CREATE TABLE IF NOT EXISTS matches (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    team1 TEXT,
    team2 TEXT,
    toss_winner TEXT,
    batting_team TEXT,
    score1 INTEGER DEFAULT 0,
    score2 INTEGER DEFAULT 0,
    winner TEXT
)
""")

cursor.execute("""
CREATE TABLE IF NOT EXISTS points (
    team TEXT PRIMARY KEY,
    played INTEGER DEFAULT 0,
    won INTEGER DEFAULT 0,
    lost INTEGER DEFAULT 0,
    points INTEGER DEFAULT 0
)
""")

conn.commit()


# ==========================================
# 1. CREATE TOURNAMENT
# ==========================================

print("\n========== CREATE TOURNAMENT ==========")

tournament_name = input("Enter Tournament Name: ")
overs = int(input("Enter Number of Overs: "))

cursor.execute(
    "INSERT INTO tournament (name, overs) VALUES (?, ?)",
    (tournament_name, overs)
)

conn.commit()

print("Tournament Created Successfully!")


# ==========================================
# 2. ADD TEAMS
# ==========================================

print("\n========== ADD TEAMS ==========")

number_of_teams = int(input("Enter Number of Teams: "))

for i in range(number_of_teams):

    team_name = input("Enter Team Name: ")
    captain = input("Enter Captain Name: ")

    try:
        cursor.execute(
            "INSERT INTO teams (name, captain) VALUES (?, ?)",
            (team_name, captain)
        )

        cursor.execute(
            "INSERT INTO points (team) VALUES (?)",
            (team_name,)
        )

        conn.commit()

        print("Team Added Successfully!")

    except sqlite3.IntegrityError:
        print("Team already exists!")


# ==========================================
# 3. ADD PLAYERS
# ==========================================

print("\n========== ADD PLAYERS ==========")

cursor.execute("SELECT id, name FROM teams")
teams = cursor.fetchall()

print("\nAvailable Teams:")

for team in teams:
    print(team[0], "-", team[1])

number_of_players = int(input("\nEnter Number of Players: "))

for i in range(number_of_players):

    player_name = input("\nEnter Player Name: ")
    team_id = int(input("Enter Team ID: "))
    role = input("Enter Role (Batsman/Bowler/All-rounder): ")

    cursor.execute("""
    INSERT INTO players
    (name, team_id, role)
    VALUES (?, ?, ?)
    """, (player_name, team_id, role))

    conn.commit()

    print("Player Added Successfully!")


# ==========================================
# 4. CREATE / SCHEDULE MATCH
# ==========================================

print("\n========== CREATE MATCH ==========")

team1 = input("Enter Team 1: ")
team2 = input("Enter Team 2: ")

cursor.execute("""
INSERT INTO matches
(team1, team2)
VALUES (?, ?)
""", (team1, team2))

match_id = cursor.lastrowid

conn.commit()

print("Match Scheduled Successfully!")


# ==========================================
# 5. TOSS
# ==========================================

print("\n========== TOSS ==========")

print("1.", team1)
print("2.", team2)

toss_choice = int(input("Enter Toss Winner (1/2): "))

if toss_choice == 1:
    toss_winner = team1
else:
    toss_winner = team2

print("Toss Winner:", toss_winner)


# ==========================================
# 6. SELECT BATTING TEAM
# ==========================================

print("\n========== SELECT BATTING TEAM ==========")

batting_team = input(
    "Enter Batting Team ("
    + team1 + " / "
    + team2 + "): "
)

cursor.execute("""
UPDATE matches
SET toss_winner = ?,
    batting_team = ?
WHERE id = ?
""", (toss_winner, batting_team, match_id))

conn.commit()


# ==========================================
# 7. ENTER SCORE
# ==========================================

print("\n========== ENTER SCORE ==========")

score1 = int(input(team1 + " Score: "))
score2 = int(input(team2 + " Score: "))

cursor.execute("""
UPDATE matches
SET score1 = ?,
    score2 = ?
WHERE id = ?
""", (score1, score2, match_id))

conn.commit()


# ==========================================
# 8. CALCULATE RESULT
# ==========================================

print("\n========== MATCH RESULT ==========")

if score1 > score2:

    winner = team1
    loser = team2

    print(winner, "won the match!")

elif score2 > score1:

    winner = team2
    loser = team1

    print(winner, "won the match!")

else:

    winner = "TIE"
    loser = ""

    print("Match Draw!")


cursor.execute("""
UPDATE matches
SET winner = ?
WHERE id = ?
""", (winner, match_id))

conn.commit()


# ==========================================
# 9. UPDATE POINTS TABLE
# ==========================================

print("\n========== UPDATE POINTS TABLE ==========")

if winner != "TIE":

    cursor.execute("""
    UPDATE points
    SET played = played + 1,
        won = won + 1,
        points = points + 2
    WHERE team = ?
    """, (winner,))

    cursor.execute("""
    UPDATE points
    SET played = played + 1,
        lost = lost + 1
    WHERE team = ?
    """, (loser,))

else:

    cursor.execute("""
    UPDATE points
    SET played = played + 1
    WHERE team IN (?, ?)
    """, (team1, team2))

conn.commit()

print("Points Table Updated!")


# ==========================================
# 10. PLAYER STATISTICS
# ==========================================

print("\n========== PLAYER STATISTICS ==========")

player_name = input("Enter Player Name: ")

runs = int(input("Enter Runs: "))
wickets = int(input("Enter Wickets: "))

cursor.execute("""
UPDATE players
SET runs = runs + ?,
    wickets = wickets + ?
WHERE name = ?
""", (runs, wickets, player_name))

conn.commit()

print("Player Statistics Updated!")


# ==========================================
# 11. GENERATE RANKING
# ==========================================

print("\n========== TEAM RANKING ==========")

cursor.execute("""
SELECT team, played, won, lost, points
FROM points
ORDER BY points DESC
""")

ranking = cursor.fetchall()

rank = 1

for row in ranking:

    print(
        rank,
        row[0],
        "| Played:", row[1],
        "| Won:", row[2],
        "| Lost:", row[3],
        "| Points:", row[4]
    )

    rank += 1


# ==========================================
# 12. FIND FINALISTS
# ==========================================

print("\n========== FINALISTS ==========")

cursor.execute("""
SELECT team
FROM points
ORDER BY points DESC
LIMIT 2
""")

finalists = cursor.fetchall()

for team in finalists:
    print(team[0])


# ==========================================
# 13. FINAL MATCH
# ==========================================

if len(finalists) >= 2:

    final_team1 = finalists[0][0]
    final_team2 = finalists[1][0]

    print("\n========== FINAL MATCH ==========")

    print(final_team1, "VS", final_team2)

    final_score1 = int(
        input(final_team1 + " Final Score: ")
    )

    final_score2 = int(
        input(final_team2 + " Final Score: ")
    )

    if final_score1 > final_score2:

        champion = final_team1

    elif final_score2 > final_score1:

        champion = final_team2

    else:

        champion = "DRAW"

else:

    champion = "Not Available"


# ==========================================
# 14. TOURNAMENT WINNER
# ==========================================

print("\n======================================")
print("       TOURNAMENT WINNER")
print("======================================")

print("Tournament:", tournament_name)

if champion != "DRAW":

    print("🏆 Winner:", champion)

else:

    print("Final Match Draw!")


# ==========================================
# END
# ==========================================

conn.close()

print("\n======================================")
print("       PROJECT COMPLETED")
print("======================================")
```

```python
import tkinter as tk
from tkinter import messagebox, ttk
import sqlite3

# ================= GUI COLOR THEME =================
# Professional cricket-style dark blue theme
BG_COLOR = "#0B1F33"
PANEL_COLOR = "#123B5D"
ACCENT_COLOR = "#2E86C1"
ACCENT_HOVER = "#3498DB"
TEXT_COLOR = "#FFFFFF"
MUTED_TEXT = "#D6EAF8"
ENTRY_BG = "#F7FBFF"
ENTRY_TEXT = "#17202A"
SUCCESS_COLOR = "#27AE60"
DANGER_COLOR = "#C0392B"


# ================= DATABASE =================

conn = sqlite3.connect("cricket.db")
cursor = conn.cursor()

cursor.execute("""
CREATE TABLE IF NOT EXISTS teams (
    team_id INTEGER PRIMARY KEY AUTOINCREMENT,
    team_name TEXT UNIQUE NOT NULL,
    captain TEXT
)
""")

cursor.execute("""
CREATE TABLE IF NOT EXISTS players (
    player_id INTEGER PRIMARY KEY AUTOINCREMENT,
    player_name TEXT NOT NULL,
    team_id INTEGER,
    role TEXT,
    jersey_no INTEGER,
    runs INTEGER DEFAULT 0,
    wickets INTEGER DEFAULT 0
)
""")

cursor.execute("""
CREATE TABLE IF NOT EXISTS matches (
    match_id INTEGER PRIMARY KEY AUTOINCREMENT,
    team1 TEXT,
    team2 TEXT,
    toss_winner TEXT,
    batting_team TEXT,
    score1 INTEGER DEFAULT 0,
    score2 INTEGER DEFAULT 0,
    winner TEXT
)
""")

cursor.execute("""
CREATE TABLE IF NOT EXISTS points (
    team_name TEXT PRIMARY KEY,
    played INTEGER DEFAULT 0,
    won INTEGER DEFAULT 0,
    lost INTEGER DEFAULT 0,
    points INTEGER DEFAULT 0
)
""")

conn.commit()

# ================= MAIN GUI =================

root = tk.Tk()

# Apply theme defaults to Tkinter widgets
root.option_add("*Font", ("Arial", 11))
root.option_add("*Background", PANEL_COLOR)
root.option_add("*Foreground", TEXT_COLOR)
root.option_add("*Label.Background", PANEL_COLOR)
root.option_add("*Label.Foreground", TEXT_COLOR)
root.option_add("*Button.Background", ACCENT_COLOR)
root.option_add("*Button.Foreground", TEXT_COLOR)
root.option_add("*Button.ActiveBackground", ACCENT_HOVER)
root.option_add("*Button.ActiveForeground", TEXT_COLOR)
root.option_add("*Entry.Background", ENTRY_BG)
root.option_add("*Entry.Foreground", ENTRY_TEXT)
root.option_add("*Entry.InsertBackground", ENTRY_TEXT)
root.option_add("*Text.Background", ENTRY_BG)
root.option_add("*Text.Foreground", ENTRY_TEXT)
root.option_add("*Text.InsertBackground", ENTRY_TEXT)
root.option_add("*OptionMenu.Background", ACCENT_COLOR)
root.option_add("*OptionMenu.Foreground", TEXT_COLOR)
root.option_add("*OptionMenu.ActiveBackground", ACCENT_HOVER)
root.option_add("*OptionMenu.ActiveForeground", TEXT_COLOR)

# ttk theme
style = ttk.Style()
try:
    style.theme_use("clam")
except tk.TclError:
    pass
style.configure("TCombobox", fieldbackground=ENTRY_BG, background=ACCENT_COLOR,
                foreground=ENTRY_TEXT, arrowcolor=TEXT_COLOR)
style.map("TCombobox", fieldbackground=[("readonly", ENTRY_BG)],
          foreground=[("readonly", ENTRY_TEXT)])
style.configure("Treeview", background=ENTRY_BG, foreground=ENTRY_TEXT,
                fieldbackground=ENTRY_BG, rowheight=30, font=("Arial", 11))
style.configure("Treeview.Heading", background=ACCENT_COLOR, foreground=TEXT_COLOR,
                font=("Arial", 11, "bold"))
style.map("Treeview", background=[("selected", ACCENT_HOVER)],
          foreground=[("selected", TEXT_COLOR)])
style.configure("TScrollbar", background=ACCENT_COLOR, troughcolor=PANEL_COLOR,
                arrowcolor=TEXT_COLOR)

root.title("Cricket Tournament Management System")
root.geometry("750x700")
root.configure(bg=BG_COLOR)

title = tk.Label(
    root,
    text="CRICKET TOURNAMENT MANAGEMENT SYSTEM",
    font=("Arial", 20, "bold"),
    bg=BG_COLOR,
    fg=TEXT_COLOR
)
title.pack(pady=(25, 5))

subtitle = tk.Label(
    root,
    text="Manage Teams • Players • Matches • Points",
    font=("Arial", 11, "italic"),
    bg=BG_COLOR,
    fg=MUTED_TEXT
)
subtitle.pack(pady=(0, 20))


# ================= ADD TEAM =================

def add_team():
    window = tk.Toplevel(root)
    window.title("Add Team")
    window.geometry("400x300")
    window.configure(bg=BG_COLOR)

    tk.Label(window, text="ADD TEAM",
             font=("Arial", 16, "bold"), bg=BG_COLOR, fg=TEXT_COLOR).pack(pady=20)

    tk.Label(window, text="Team Name").pack()
    team_name = tk.Entry(window)
    team_name.pack(pady=5)

    tk.Label(window, text="Captain Name").pack()
    captain = tk.Entry(window)
    captain.pack(pady=5)

    def save_team():
        name = team_name.get().strip()
        cap = captain.get().strip()

        if not name or not cap:
            messagebox.showerror("Error", "Please enter all details")
            return

        try:
            cursor.execute(
                "INSERT INTO teams (team_name, captain) VALUES (?, ?)",
                (name, cap)
            )

            cursor.execute(
                "INSERT OR IGNORE INTO points (team_name) VALUES (?)",
                (name,)
            )

            conn.commit()
            messagebox.showinfo("Success", "Team Added Successfully!")
            window.destroy()

        except sqlite3.IntegrityError:
            messagebox.showerror("Error", "Team already exists!")

    tk.Button(window, text="ADD TEAM",
              command=save_team, width=20,
              bg=SUCCESS_COLOR, fg=TEXT_COLOR,
              activebackground="#2ECC71", activeforeground=TEXT_COLOR,
              relief="flat", cursor="hand2").pack(pady=20)


# ================= ADD PLAYER =================

def add_player():
    window = tk.Toplevel(root)
    window.title("Add Player")
    window.geometry("450x450")
    window.configure(bg=BG_COLOR)

    tk.Label(window, text="ADD PLAYER",
             font=("Arial", 16, "bold"), bg=BG_COLOR, fg=TEXT_COLOR).pack(pady=15)

    tk.Label(window, text="Player Name").pack()
    player_name = tk.Entry(window)
    player_name.pack()

    tk.Label(window, text="Team ID").pack()
    team_id = tk.Entry(window)
    team_id.pack()

    tk.Label(window, text="Role").pack()

    role = tk.StringVar(value="Batsman")
    roles = ["Batsman", "Bowler", "All-Rounder", "Wicket Keeper"]

    tk.OptionMenu(window, role, *roles).pack()

    tk.Label(window, text="Jersey Number").pack()
    jersey = tk.Entry(window)
    jersey.pack()

    def save_player():
        name = player_name.get().strip()

        try:
            tid = int(team_id.get())
            jno = int(jersey.get())
        except ValueError:
            messagebox.showerror(
                "Error",
                "Team ID and Jersey Number must be numbers"
            )
            return

        if not name:
            messagebox.showerror("Error", "Enter Player Name")
            return

        cursor.execute("""
        INSERT INTO players
        (player_name, team_id, role, jersey_no)
        VALUES (?, ?, ?, ?)
        """, (name, tid, role.get(), jno))

        conn.commit()

        messagebox.showinfo("Success", "Player Added Successfully!")
        window.destroy()

    tk.Button(window, text="ADD PLAYER",
              command=save_player, width=20,
              bg=SUCCESS_COLOR, fg=TEXT_COLOR,
              activebackground="#2ECC71", activeforeground=TEXT_COLOR,
              relief="flat", cursor="hand2").pack(pady=20)


# ================= VIEW PLAYERS =================

def view_players():
    window = tk.Toplevel(root)
    window.title("Player Statistics")
    window.geometry("800x500")
    window.configure(bg=BG_COLOR)

    text = tk.Text(window, font=("Courier New", 11), bg=ENTRY_BG, fg=ENTRY_TEXT, insertbackground=ENTRY_TEXT)
    text.pack(fill="both", expand=True)

    cursor.execute("""
    SELECT players.player_name,
           teams.team_name,
           players.role,
           players.jersey_no,
           players.runs,
           players.wickets
    FROM players
    LEFT JOIN teams
    ON players.team_id = teams.team_id
    """)

    players = cursor.fetchall()

    text.insert(tk.END, "PLAYER STATISTICS\n")
    text.insert(tk.END, "=" * 80 + "\n")

    for p in players:
        text.insert(
            tk.END,
            f"Name: {p[0]}\n"
            f"Team: {p[1]}\n"
            f"Role: {p[2]}\n"
            f"Jersey: {p[3]}\n"
            f"Runs: {p[4]}\n"
            f"Wickets: {p[5]}\n"
            + "-" * 80 + "\n"
        )


# ================= VIEW TEAMS =================

def view_teams():
    window = tk.Toplevel(root)
    window.title("Teams")
    window.geometry("600x400")
    window.configure(bg=BG_COLOR)

    text = tk.Text(window, font=("Arial", 12), bg=ENTRY_BG, fg=ENTRY_TEXT, insertbackground=ENTRY_TEXT)
    text.pack(fill="both", expand=True)

    cursor.execute("SELECT * FROM teams")
    teams = cursor.fetchall()

    text.insert(tk.END, "TEAMS\n\n")

    for team in teams:
        text.insert(
            tk.END,
            f"Team ID: {team[0]}\n"
            f"Team Name: {team[1]}\n"
            f"Captain: {team[2]}\n"
            + "-" * 40 + "\n"
        )


# ================= UPDATE POINTS =================

def update_points():
    window = tk.Toplevel(root)
    window.title("Update Points Table")
    window.geometry("500x400")
    window.configure(bg=BG_COLOR)

    tk.Label(
        window,
        text="UPDATE POINTS TABLE",
        font=("Arial", 17, "bold"),
        bg=BG_COLOR,
        fg=TEXT_COLOR
    ).pack(pady=20)

    cursor.execute("SELECT team_name FROM points ORDER BY team_name")
    teams = [row[0] for row in cursor.fetchall()]

    if len(teams) < 2:
        messagebox.showerror(
            "Error",
            "At least 2 teams are required to update points."
        )
        window.destroy()
        return

    tk.Label(window, text="Select Winner").pack(pady=5)

    winner_var = tk.StringVar()
    winner_menu = ttk.Combobox(
        window,
        textvariable=winner_var,
        values=teams,
        state="readonly",
        width=30
    )
    winner_menu.pack()

    tk.Label(window, text="Select Loser").pack(pady=5)

    loser_var = tk.StringVar()
    loser_menu = ttk.Combobox(
        window,
        textvariable=loser_var,
        values=teams,
        state="readonly",
        width=30
    )
    loser_menu.pack()

    tk.Label(
        window,
        text="Winner gets +2 points",
        font=("Arial", 10)
    ).pack(pady=10)

    def save_points():
        winner = winner_var.get()
        loser = loser_var.get()

        if not winner or not loser:
            messagebox.showerror(
                "Error",
                "Please select both winner and loser."
            )
            return

        if winner == loser:
            messagebox.showerror(
                "Error",
                "Winner and loser must be different."
            )
            return

        # Winner update
        cursor.execute("""
        UPDATE points
        SET played = played + 1,
            won = won + 1,
            points = points + 2
        WHERE team_name = ?
        """, (winner,))

        # Loser update
        cursor.execute("""
        UPDATE points
        SET played = played + 1,
            lost = lost + 1
        WHERE team_name = ?
        """, (loser,))

        conn.commit()

        messagebox.showinfo(
            "Success",
            f"Points updated successfully!\n\n"
            f"Winner: {winner} (+2 points)\n"
            f"Loser: {loser}"
        )

        window.destroy()


    tk.Button(
        window,
        text="UPDATE POINTS",
        command=save_points,
        width=25,
        height=2,
        bg=SUCCESS_COLOR,
        fg=TEXT_COLOR,
        activebackground="#2ECC71",
        activeforeground=TEXT_COLOR,
        relief="flat",
        cursor="hand2"
    ).pack(pady=20)


# ================= POINTS TABLE =================

def points_table():
    window = tk.Toplevel(root)
    window.title("View Points Table")
    window.geometry("900x550")
    window.configure(bg=BG_COLOR)

    tk.Label(
        window,
        text="CRICKET TOURNAMENT - POINTS TABLE",
        font=("Arial", 20, "bold"),
        bg=BG_COLOR,
        fg=TEXT_COLOR
    ).pack(pady=15)

    frame = tk.Frame(window)
    frame.pack(fill="both", expand=True, padx=20, pady=10)

    columns = ("Rank", "Team", "Played", "Won", "Lost", "Points")

    table = ttk.Treeview(
        frame,
        columns=columns,
        show="headings",
        height=15
    )

    for col in columns:
        table.heading(col, text=col)

    table.column("Rank", width=80, anchor="center")
    table.column("Team", width=300, anchor="w")
    table.column("Played", width=110, anchor="center")
    table.column("Won", width=110, anchor="center")
    table.column("Lost", width=110, anchor="center")
    table.column("Points", width=110, anchor="center")

    scrollbar = ttk.Scrollbar(
        frame,
        orient="vertical",
        command=table.yview
    )

    table.configure(yscrollcommand=scrollbar.set)

    table.pack(side="left", fill="both", expand=True)
    scrollbar.pack(side="right", fill="y")

    def load_points():
        # Clear old data
        for item in table.get_children():
            table.delete(item)

        # Get latest data from database
        cursor.execute("""
        SELECT team_name, played, won, lost, points
        FROM points
        ORDER BY points DESC, won DESC, team_name ASC
        """)

        data = cursor.fetchall()

        # Display updated points table
        for rank, row in enumerate(data, start=1):
            table.insert(
                "",
                "end",
                values=(
                    rank,
                    row[0],
                    row[1],
                    row[2],
                    row[3],
                    row[4]
                )
            )

    def close_table():
        window.destroy()

    # Load current points immediately
    load_points()

    button_frame = tk.Frame(window)
    button_frame.pack(pady=15)

    tk.Button(
        button_frame,
        text="REFRESH",
        command=load_points,
        width=18,
        height=2,
        bg=ACCENT_COLOR,
        fg=TEXT_COLOR,
        activebackground=ACCENT_HOVER,
        activeforeground=TEXT_COLOR,
        relief="flat",
        cursor="hand2"
    ).pack(side="left", padx=10)

    tk.Button(
        button_frame,
        text="CLOSE",
        command=close_table,
        width=18,
        height=2,
        bg=DANGER_COLOR,
        fg=TEXT_COLOR,
        activebackground="#E74C3C",
        activeforeground=TEXT_COLOR,
        relief="flat",
        cursor="hand2"
    ).pack(side="left", padx=10)


# ================= BUTTONS =================

def create_main_button(text, command, pady=5):
    button = tk.Button(
        root,
        text=text,
        command=command,
        width=30,
        height=2,
        font=("Arial", 12, "bold"),
        bg=ACCENT_COLOR,
        fg=TEXT_COLOR,
        activebackground=ACCENT_HOVER,
        activeforeground=TEXT_COLOR,
        relief="flat",
        cursor="hand2"
    )
    button.pack(pady=pady)
    return button

create_main_button("ADD TEAM", add_team)
create_main_button("ADD PLAYER", add_player)
create_main_button("VIEW TEAMS", view_teams)
create_main_button("PLAYER STATISTICS", view_players)
create_main_button("UPDATE POINTS", update_points)
create_main_button("VIEW POINTS TABLE", points_table)
create_main_button("EXIT", root.destroy, pady=20)


# ================= START GUI =================

root.mainloop()
```
